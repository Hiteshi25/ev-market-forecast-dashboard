```python
# PART 1: Import libraries and check dataset path

import pandas as pd
import os

file_path = "../data/raw/EV data by country 2026.xlsx"

print("File exists:",os.path.exists(file_path))
```

```python
# PART 2: Load the EV dataset

df = pd.read_excel(file_path,sheet_name="GEVO_EV_2026")

print("Dataset loaded successfully")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
df.head()
```

```python
# PART 9: Check duplicate rows

duplicates = df[df.duplicated(keep=False)]

print(duplicates)
```

```python
# PART 15: Select relevant parameters for the EV market project

selected_parameters = [
    "EV sales",
    "EV stock",
    "EV sales share",
    "EV stock share",
    "EV charging points",
    "Electric LDVs per charging point",
    "Battery deployment"
]

df_clean = df[df["parameter"].isin(selected_parameters)].copy()

print("Rows after parameter selection:",df_clean.shape[0])
print("Columns:",df_clean.shape[1])
```

```python
# PART 17: Remove rows with missing region or country

df_clean = df_clean.dropna(subset=["region_country"])

print("Rows after removing missing regions:",df_clean.shape[0])
```

```python
# PART 39: Create historical EV sales share dataset

ev_sales_share = df_clean[
    (df_clean["parameter"] == "EV sales share") &
    (df_clean["category"] == "Historical")
].copy()

print("Rows:",ev_sales_share.shape[0])
print(ev_sales_share[["region_country","mode","powertrain","year","unit","value"]].head(20))
```

```python
# PART 40: Create global EV sales share trend

global_sales_share = ev_sales_share[
    (ev_sales_share["region_country"] == "World") &
    (ev_sales_share["powertrain"] == "EV")
].copy()

global_sales_share = (
    global_sales_share
    .groupby("year",as_index=False)["value"]
    .mean()
    .sort_values("year")
)

print(global_sales_share)
```

```python
# PART 41: Visualize global EV sales share

fig = px.line(
    global_sales_share,
    x="year",
    y="value",
    markers=True,
    title="Global EV Sales Share (2010–2025)",
    labels={
        "year":"Year",
        "value":"EV Sales Share (%)"
    }
)

fig.update_layout(
    width=1200,
    height=600,
    paper_bgcolor="black",
    plot_bgcolor="black",
    font=dict(color="white"),
    title_font=dict(color="white",size=22),
    xaxis=dict(
        title="Year",
        color="white",
        gridcolor="gray"
    ),
    yaxis=dict(
        title="EV Sales Share (%)",
        color="white",
        gridcolor="gray"
    ),
    hovermode="x unified"
)

fig.show()
```

```python
# PART 42: Inspect global EV sales share by mode and powertrain

global_share_check = ev_sales_share[
    ev_sales_share["region_country"] == "World"
].copy()

print(
    global_share_check[
        ["mode","powertrain","year","unit","value"]
    ].to_string(index=False)
)
```

```python
# PART 43: Calculate weighted global EV sales share correctly

global_sales = ev_sales[
    (ev_sales["region_country"] == "World") &
    (ev_sales["powertrain"] == "EV")
][["mode","year","value"]].copy()

global_sales = global_sales.rename(columns={"value":"ev_sales"})

global_share = ev_sales_share[
    (ev_sales_share["region_country"] == "World") &
    (ev_sales_share["powertrain"] == "EV")
][["mode","year","value"]].copy()

global_share = global_share.rename(columns={"value":"ev_sales_share"})

global_share = global_share[global_share["ev_sales_share"] > 0]

global_share_data = pd.merge(
    global_sales,
    global_share,
    on=["mode","year"],
    how="inner"
)

global_share_data["total_vehicle_sales"] = (
    global_share_data["ev_sales"] /
    (global_share_data["ev_sales_share"] / 100)
)

global_share_data = (
    global_share_data
    .groupby("year",as_index=False)
    .agg(
        ev_sales=("ev_sales","sum"),
        total_vehicle_sales=("total_vehicle_sales","sum")
    )
)

global_share_data["ev_sales_share"] = (
    global_share_data["ev_sales"] /
    global_share_data["total_vehicle_sales"] * 100
)

print(global_share_data)
```

```python
# PART 44: Visualize corrected global EV sales share

fig = px.line(
    global_share_data,
    x="year",
    y="ev_sales_share",
    markers=True,
    title="Global EV Sales Share (2010–2025)",
    labels={
        "year":"Year",
        "ev_sales_share":"EV Sales Share (%)"
    }
)

fig.update_layout(
    width=1200,
    height=600,
    paper_bgcolor="black",
    plot_bgcolor="black",
    font=dict(color="white"),
    title_font=dict(color="white",size=22),
    xaxis=dict(
        title="Year",
        color="white",
        gridcolor="gray"
    ),
    yaxis=dict(
        title="EV Sales Share (%)",
        color="white",
        gridcolor="gray"
    ),
    hovermode="x unified"
)

fig.show()
```
